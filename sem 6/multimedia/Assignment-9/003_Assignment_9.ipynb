```python
# Import required libraries for data processing, image handling, and performance measurement
import numpy as np
import pandas as pd
import time
from collections import Counter
from math import log2
from PIL import Image
from google.colab import files
```

```python
# Define Shannon–Fano coding algorithm along with encoding and entropy calculation functions
def shannon_fano(symbols):
    codes = {symbol: "" for symbol in symbols}

    def split(symbol_list):
        if len(symbol_list) <= 1:
            return

        total = sum(freq for _, freq in symbol_list)
        acc = 0

        for i in range(len(symbol_list)):
            acc += symbol_list[i][1]
            if acc >= total / 2:
                break

        left = symbol_list[:i+1]
        right = symbol_list[i+1:]

        for symbol, _ in left:
            codes[symbol] += "0"
        for symbol, _ in right:
            codes[symbol] += "1"

        split(left)
        split(right)

    sorted_symbols = sorted(symbols.items(), key=lambda x: x[1], reverse=True)
    split(sorted_symbols)

    return codes


def encode_data(data, codes):
    return ''.join(codes[symbol] for symbol in data)


def entropy_calc(prob):
    return -sum(p * log2(p) for p in prob if p > 0)


def process_file(file_name, file_bytes):
    start = time.time()

    # 📄 TEXT FILE
    if file_name.endswith(".txt"):
        text = file_bytes.decode("utf-8")
        data = list(text)
        data_type = "Text"

    # 🖼️ IMAGE FILE
    else:
        import io
        img = Image.open(io.BytesIO(file_bytes)).convert('L')
        data = list(np.array(img).flatten())
        data_type = "Image"

    freq = Counter(data)
    codes = shannon_fano(freq)
    encoded = encode_data(data, codes)

    total = len(data)
    prob = {k: v/total for k, v in freq.items()}

    entropy = entropy_calc(prob.values())
    avg_len = sum(prob[s] * len(codes[s]) for s in codes)

    input_size = total * 8
    compressed_size = len(encoded)
    compression_ratio = input_size / compressed_size if compressed_size else 0
    efficiency = entropy / avg_len if avg_len else 0

    end = time.time()

    return {
        "File Name": file_name,
        "Type": data_type,
        "Input Size (bits)": input_size,
        "Compressed Size (bits)": compressed_size,
        "Compression Ratio": compression_ratio,
        "Average Code Length": avg_len,
        "Entropy": entropy,
        "Efficiency": efficiency,
        "Compression Time (s)": end - start
    }
```

```python
# Initialize a list to store results of multiple file compressions
results = []
```

```python
# Convert collected results into a structured table using pandas DataFrameuploaded = files.upload()
for file_name in uploaded:
    file_bytes = uploaded[file_name]
    result = process_file(file_name, file_bytes)
    results.append(result)

    print(f"\nProcessed: {file_name}")
    for k, v in result.items():
        if k not in ["File Name", "Type"]:
            print(f"{k}: {v}")
```

```python
# Save the final results table as a CSV file for submission and analysis
df = pd.DataFrame(results)
df
```
